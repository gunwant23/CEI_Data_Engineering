## 1. Load the CSV dataset into a Pandas DataFrame

```python
import pandas as pd
import numpy as np

# The file uses Latin-1 encoding (contains extended-ASCII characters)
df = pd.read_csv('Sample_-_Superstore.csv', encoding='latin1')
print("Dataset loaded successfully.")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

## 3. Handle missing values
First identify how many missing values exist per column, then decide how to handle them
(fill with a sensible default, or drop rows/columns as appropriate).

```python
# Count of missing values per column
missing_counts = df.isnull().sum()
missing_counts[missing_counts > 0] if missing_counts.sum() > 0 else "No missing values found in the dataset."
```

```python
# This dataset happens to be complete (no missing values), but here is the general
# pattern used to handle missing data:

# Option A: Drop rows where ALL values are missing
df = df.dropna(how='all')

# Option B: Fill missing numeric values with the column median (example, applied defensively)
numeric_cols = df.select_dtypes(include=[np.number]).columns
for c in numeric_cols:
    if df[c].isnull().any():
        df[c] = df[c].fillna(df[c].median())

# Option C: Fill missing text/categorical values with 'Unknown'
text_cols = df.select_dtypes(include=['object']).columns
for c in text_cols:
    if df[c].isnull().any():
        df[c] = df[c].fillna('Unknown')

print("Missing values after cleaning:", df.isnull().sum().sum())
```

## 4. Perform basic operations (filter rows, select columns)

```python
# Select a subset of columns
subset = df[['Order ID', 'Customer Name', 'Category', 'Sales', 'Quantity', 'Profit']]
subset.head()
```

```python
# Filter rows: orders in the 'Furniture' category with Sales > 500
furniture_high_sales = df[(df['Category'] == 'Furniture') & (df['Sales'] > 500)]
print(f"Found {len(furniture_high_sales)} Furniture orders with Sales > 500")
furniture_high_sales.head()
```

```python
# Filter rows: profitable orders shipped via 'Same Day'
same_day_profitable = df[(df['Ship Mode'] == 'Same Day') & (df['Profit'] > 0)]
print(f"Found {len(same_day_profitable)} profitable Same Day orders")
same_day_profitable.head()
```

## 5. Remove duplicates

```python
# Check for fully duplicated rows
dup_count = df.duplicated().sum()
print(f"Duplicate rows found: {dup_count}")

# Remove them (keep the first occurrence)
df = df.drop_duplicates(keep='first')
print(f"Rows after removing duplicates: {len(df)}")
```

## 6. Create a derived column: `total_amount = price * quantity`

The raw dataset doesn't include a standalone "price per unit" column — it has `Sales`
(the total sale amount for that line) and `Quantity`. To follow the exercise exactly, we:

1. Derive a `price` column as the unit price (`Sales / Quantity`)
2. Compute `total_amount = price * quantity`

As a sanity check, `total_amount` should equal the original `Sales` column.

```python
# Derive unit price
df['price'] = df['Sales'] / df['Quantity']

# Derived column: total_amount = price * quantity
df['total_amount'] = df['price'] * df['Quantity']

# Sanity check: total_amount should match Sales (within floating point rounding)
df['sales_matches_total'] = np.isclose(df['total_amount'], df['Sales'], atol=0.01)
print("Rows where total_amount matches Sales:", df['sales_matches_total'].sum(), "/", len(df))

df[['Sales', 'Quantity', 'price', 'total_amount']].head()
```

```python
# Drop the helper verification column before saving
df = df.drop(columns=['sales_matches_total'])
```

## 7. Save the cleaned dataset as a new CSV file

```python
output_path = 'Sample_Superstore_cleaned.csv'
df.to_csv(output_path, index=False)
print(f"Cleaned dataset saved to: {output_path}")
print(f"Final shape: {df.shape}")
```
